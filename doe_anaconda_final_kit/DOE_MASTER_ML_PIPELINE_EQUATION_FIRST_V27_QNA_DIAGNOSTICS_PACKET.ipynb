## What V27 Checks

- Stability: whether interval stability columns already exist, whether public stability rows match, and whether a pressure-temperature phase-curve check can run.

- Feature families: which well is driven by resistivity, elastic, lithology, and density/porosity evidence.

```python
from pathlib import Path
import os
import subprocess
import sys

REPO_ROOT = Path.cwd()
SCRIPT = REPO_ROOT / "doe_anaconda_final_kit" / "run_qna_diagnostics_v27.py"

# Edit these paths on the DOE laptop if the environment variables are not set.
MODEL_MATRIX = os.environ.get(
    "QNA_MODEL_MATRIX",
    r"C:\\Users\\rohan.nanda\\Downloads\\outputs_runtime\\ann_loo_scatter_plots\\ann_loo_model_matrix_v26_ann_loo_scatter_packet.csv",
)
V26_OUTPUT_DIR = os.environ.get("QNA_V26_OUTPUT_DIR", "")
IMPORTANCE_CSV = os.environ.get("QNA_IMPORTANCE_CSV", "")
OUTPUT_DIR = os.environ.get(
    "QNA_OUTPUT_DIR",
    str(REPO_ROOT / "outputs_runtime" / "qna_diagnostics_v27" / "notebook_run"),
)
PERSONAL_REVIEW_EMAIL = os.environ.get("PERSONAL_REVIEW_EMAIL", "")

print("Script:", SCRIPT)
print("Model matrix:", MODEL_MATRIX)
print("V26 output dir:", V26_OUTPUT_DIR or "auto-detect latest repo outputs_runtime/ann_loo_scatter_plots")
print("Importance CSV:", IMPORTANCE_CSV or "optional / not set")
print("Output dir:", OUTPUT_DIR)
print("Email recipient:", PERSONAL_REVIEW_EMAIL or "blank Outlook draft recipient")
```
