# DOE V20 Three Fixed Versions Launcher

This lightweight launcher runs the Git-safe V20 comparison scaffold:

- V20A baseline safe-normalized Ridge/logistic reference.
- V20B reversed-weight equation-dominance stress test.
- V20C HYDRATE-02 core-prior auxiliary check.

Private model matrices, row-level predictions, fitted models, and runtime logs stay in ignored runtime folders.

```python
from pathlib import Path
import os
import subprocess
import sys

repo_root = Path.cwd()
if repo_root.name == "v20_three_versions":
    repo_root = repo_root.parents[1]
elif repo_root.name == "doe_anaconda_final_kit":
    repo_root = repo_root.parent

script = repo_root / "doe_anaconda_final_kit" / "v20_three_versions" / "run_v20_three_versions.py"
print("Runner:", script)

# Set this to the private standardized model matrix exported in the DOE runtime.
# Required columns: well_alias and hydrate_saturation_reference.
MODEL_MATRIX_CSV = os.environ.get("V20_MODEL_MATRIX_CSV", r"")

# HYDRATE-02 candidate target rows are ignored runtime data. Default path is used if this is blank.
CORE_CANDIDATE_CSV = os.environ.get("V20_CORE_CANDIDATE_CSV", r"")

cmd = [sys.executable, str(script), "--print-plan"]
if MODEL_MATRIX_CSV:
    cmd.extend(["--model-matrix-csv", MODEL_MATRIX_CSV])
if CORE_CANDIDATE_CSV:
    cmd.extend(["--core-candidate-csv", CORE_CANDIDATE_CSV])

print("Command:", " ".join(cmd))
subprocess.run(cmd, check=True)
```
